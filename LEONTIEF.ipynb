```python
import pandas as pd
import numpy as np
import pymrio
from pathlib import Path
import gc

EXIOBASE_file = Path(
    r"C:\Users\menne\Documents\Master\Thesis\Data\MRIO\Exiobase\IOT_2020_pxp.zip"
)

L_file = Path(
    r"C:\Users\menne\Documents\Master\Thesis\Data\MRIO\Exiobase\L_2020_pxp.pkl"
)

# Load EXIOBASE
exio = pymrio.parse_exiobase3(EXIOBASE_file)

print("EXIOBASE loaded")

# ---------------------------------------------------------
# Technical coefficient matrix
# A = Z * diag(x)^-1
# ---------------------------------------------------------

print("Calculating A...")
A = pymrio.calc_A(exio.Z, exio.x)

print("A calculated:", A.shape)
print("A memory [GB]:", A.memory_usage(deep=True).sum() / 1e9)

# EXIOBASE object is no longer required
del exio
gc.collect()

print("EXIOBASE object removed from memory")

# ---------------------------------------------------------
# Full Leontief inverse
# L = (I - A)^-1
# ---------------------------------------------------------

print("Calculating full Leontief inverse...")

L = pymrio.calc_L(A)

print("L calculated:", L.shape)
print("L memory [GB]:", L.memory_usage(deep=True).sum() / 1e9)

# ---------------------------------------------------------
# Save L
# ---------------------------------------------------------

L.to_pickle(L_file)

print("Leontief inverse saved to:")
print(L_file)

```
